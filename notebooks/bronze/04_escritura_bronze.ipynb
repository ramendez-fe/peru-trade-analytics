```python
from pyspark.sql import functions as F

def escribir_bronze(df_spark, fuente: str, nombre_tabla: str):
    """
    Escribe un DataFrame en Delta bajo bronze, con metadatos
    de trazabilidad, y registra la carga en bronze.control_ingestas.
    """
    # 1. Agregamos las columnas de auditoría
    df_final = (
        df_spark
        .withColumn("fecha_extraccion", F.current_timestamp())
        .withColumn("fuente", F.lit(fuente))
    )
    
    # 2. Guardamos la tabla principal
    df_final.write.format("delta").mode("overwrite") \
        .option("overwriteSchema", "true") \
        .saveAsTable(f"bronze.{nombre_tabla}")
        
    # 3. Creamos el registro para la tabla de control
    registros = df_final.count()
    control_row = spark.createDataFrame(
        [(fuente, registros)],
        ["fuente", "registros_obtenidos"]
    ).withColumn("fecha_extraccion", F.current_timestamp())
    
    # 4. Guardamos en la tabla de control (usamos 'append' para mantener el historial)
    control_row.write.format("delta").mode("append") \
        .saveAsTable("bronze.control_ingestas")
        
    print(f"{fuente}: {registros} registros escritos en bronze.{nombre_tabla}")
```

```python
# Ejecutamos para nuestras 4 tablas limpias
escribir_bronze(df_g6_spark, fuente="sunat", nombre_tabla="sunat_g6")
escribir_bronze(df_g9_spark, fuente="sunat", nombre_tabla="sunat_g9")
escribir_bronze(df_comtrade_spark, fuente="comtrade", nombre_tabla="comtrade")
escribir_bronze(df_bcrp_spark, fuente="bcrp", nombre_tabla="tipo_cambio")
```
